**Data Loading and EDA**

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
#Load the dataset
df = pd.read_csv('/content/mushrooms.csv')
df
```

```python
#Show all the column names
print(df.columns)
```

```python
#Display the dataset information
df.info()
```

```python
df.describe()
```

```python
# Display the unique values in each column
for col in df.columns:
    print(f"Unique values in '{col}': {df[col].unique()}")
```

```python
#Replace the '?' with null and count # of null
df1=df.replace('?', np.nan)
print(df1.isnull().sum())
```

There is a lot of missing values in stalk-root, which can not be replaced by average value --> eliminate this column. Also, eliminate veil-type because it only has 1 value --> not meaningful for classification

```python
#Drop stalk-root and veil-type columns
df2=df1.drop(['stalk-root', 'veil-type'], axis=1)
df2
```

```python
#Visualize the distribution of 2 classes
sns.countplot(x='class', width=0.3, data=df2)
plt.title('Distribution of Classes')
plt.show()
```

```python
# Visualize frequency of classes by value of features
for col in df2.columns:
    if col != 'class':
        plt.figure()
        sns.countplot(x=col, hue='class', data=df2, palette='pastel')
        plt.title(f'Frequency of {col} values by class')
        plt.show()
```

The distribution of the values in gill-attachment, gill-spacing, stalk-shape, veil-color and ring-number are considered to be highly skewed and the class distribution in those values is almost equal --> not contribute much values in the classification process --> drop

```python
#Drop gill-attachment, gill-spacing, veil-color, ring-number, stalk-shape, stalk-surface-below-ring, stalk-color-below-ring
df2=df2.drop(['gill-attachment', 'gill-spacing', 'veil-color', 'ring-number','stalk-shape','stalk-surface-below-ring','stalk-color-below-ring'], axis=1)
df2
```

**Data Preprocessing**

```python
#Define features (X) and target (y)
X = df2.drop('class', axis=1)
y = df2['class']
```

```python
#Encode and fit the y-label (0=e, 1=p)
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(df2['class'])
print(y_encoded)
```

```python
#Split data into training and testing sets
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.3, random_state=42)
```

```python
#Encode and fit X_train
from sklearn.preprocessing import OneHotEncoder
encoder=OneHotEncoder(drop="first",sparse_output=False)
X_train_encoded=encoder.fit_transform(X_train)
print(X_train_encoded)
```

```python
#Encode X_test
X_test_encoded=encoder.transform(X_test)
print(X_test_encoded)
```

**Model Training and Evaluation**

```python
#Gaussian Naive Bayes
from sklearn.naive_bayes import GaussianNB
```

```python
gnb_model=GaussianNB()
gnb_model.fit(X_train_encoded, y_train)
```

```python
y_pred_gnb=gnb_model.predict(X_test_encoded)
print(y_pred_gnb)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred_gnb))
```

```python
#Random Forest
from sklearn.ensemble import RandomForestClassifier
```

```python
rfc_model=RandomForestClassifier()
rfc_model.fit(X_train_encoded, y_train)
```

```python
y_pred_rfc=rfc_model.predict(X_test_encoded)
print(y_pred_rfc)
```

```python
print(classification_report(y_test, y_pred_rfc))
```

```python
#Decision Tree
from sklearn.tree import DecisionTreeClassifier
```

```python
dtc_model=DecisionTreeClassifier()
dtc_model.fit(X_train_encoded, y_train)
```

```python
y_pred_dtc=dtc_model.predict(X_test_encoded)
print(y_pred_dtc)
```

```python
print(classification_report(y_test, y_pred_dtc))
```

```python
#Logistic Regression
from sklearn.linear_model import LogisticRegression
```

```python
lr_model=LogisticRegression()
lr_model.fit(X_train_encoded, y_train)
```

```python
y_pred_lr=lr_model.predict(X_test_encoded)
print(y_pred_lr)
```

```python
print(classification_report(y_test, y_pred_lr))
```

```python
#Support Vector Classication (SVC)
from sklearn.svm import SVC
```

```python
svc_model=SVC()
svc_model.fit(X_train_encoded, y_train)
```

```python
y_pred_svc=svc_model.predict(X_test_encoded)
print(y_pred_svc)
```

```python
print(classification_report(y_test, y_pred_svc))
```

```python
#K-Nearest Neighbors
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn_model=KNeighborsClassifier()
knn_model.fit(X_train_encoded, y_train)
```

```python
y_pred_knn=knn_model.predict(X_test_encoded)
print(y_pred_knn)
```

```python
print(classification_report(y_test, y_pred_knn))
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
# Store predictions in a dictionary
y_preds = {
    'gnb': y_pred_gnb,
    'rfc': y_pred_rfc,
    'dtc': y_pred_dtc,
    'lr': y_pred_lr,
    'svc': y_pred_svc,
    'knn': y_pred_knn,
    'xgb': y_pred_xgb,
}

# Initialize dictionaries to store metrics
acc = {}
pre = {}
rec = {}
f1 = {}

# Calculate metrics for each model
for model_name, y_pred in y_preds.items():
    acc[model_name] = accuracy_score(y_test, y_pred)
    pre[model_name] = precision_score(y_test, y_pred)
    rec[model_name] = recall_score(y_test, y_pred)
    f1[model_name] = f1_score(y_test, y_pred)
```

```python
#Create a dataframe for a table view
model_comparison={
    'Model':['Gaussian Naive Bayer','Random Forest','Decision Tree','Logistic Regression','Support Vector Machine','K-Nearest Neighbors','XGBoost'],
    'Accuracy': list(acc.values()),
    'F1-Score': list(f1.values()),
    'Precision': list(pre.values()),
    'Recall': list(rec.values())
}
df_model_comparison=pd.DataFrame(model_comparison).round(3)
df_model_comparison
```
